```python
def load_data(data_path, n_files=None):

    """
    Load parquet files into one DataFrame.
    """

    files = get_parquet_files(data_path, n_files)

    dfs = []

    for i, file in enumerate(files, start=1):

        print(f"[{i}/{len(files)}] Reading {file.name}")

        temp = pd.read_parquet(file)

        temp["Source_File"] = file.name

        dfs.append(temp)

    print("Concatenating...")

    df = pd.concat(dfs, ignore_index=True)

    print("Done!")
    print(f"Shape: {df.shape}")

    return df
```

```python
def create_combined_daily_dataset(df, remove_returns=True):
    """
    Create daily aggregated dataset for the Combined channel.

    Parameters
    ----------
    df : pd.DataFrame
        Raw transaction data.

    remove_returns : bool, default=True
        If True, remove rows with negative/zero sales or quantity.

    Returns
    -------
    pd.DataFrame
        Daily aggregated dataset.
    """

    data = df.copy()

    # ---------------------------------------
    # Convert Date
    # ---------------------------------------
    data["transaction_date"] = pd.to_datetime(data["transaction_date"])
    # ---------------------------------------
    # Convert Sales & Quantity Sign
    # ---------------------------------------
    data["net_amount"] = data["net_amount"] * -1
    data["quantity"] = data["quantity"] * -1
    if remove_returns:

        data = data[
            (data["net_amount"] > 0) &
            (data["quantity"] > 0)
        ].copy()

    # ---------------------------------------
    # Total Sales
    # ---------------------------------------
    total_sales = (
        data.groupby("transaction_date")["net_amount"]
        .sum()
        .rename("Total Sales")
    )

    # ---------------------------------------
    # Loyal Sales
    # ---------------------------------------
    loyal_sales = (
        data[data["is_loyal"]]
        .groupby("transaction_date")["net_amount"]
        .sum()
        .rename("Loyal Sales")
    )

    # ---------------------------------------
    # Non Loyal Sales
    # ---------------------------------------
    non_loyal_sales = (
        data[~data["is_loyal"]]
        .groupby("transaction_date")["net_amount"]
        .sum()
        .rename("Non Loyal Sales")
    )

    # ---------------------------------------
    # Total Transactions
    # ---------------------------------------
    total_transactions = (
        data.groupby("transaction_date")["receipt_no"]
        .nunique()
        .rename("Total Transactions")
    )

    # ---------------------------------------
    # Loyal Transactions
    # ---------------------------------------
    loyal_transactions = (
        data[data["is_loyal"]]
        .groupby("transaction_date")["receipt_no"]
        .nunique()
        .rename("Loyal Transactions")
    )

    # ---------------------------------------
    # Non Loyal Transactions
    # ---------------------------------------
    non_loyal_transactions = (
        data[~data["is_loyal"]]
        .groupby("transaction_date")["receipt_no"]
        .nunique()
        .rename("Non Loyal Transactions")
    )

    # ---------------------------------------
    # Item Quantity
    # ---------------------------------------
    item_quantity = (
        data.groupby("transaction_date")["quantity"]
        .sum()
        .rename("Item Quantity")
    )

    # ---------------------------------------
    # Combine All Metrics
    # ---------------------------------------
    daily = pd.concat([
        total_sales,
        loyal_sales,
        non_loyal_sales,
        total_transactions,
        loyal_transactions,
        non_loyal_transactions,
        item_quantity
    ], axis=1).fillna(0)

    # ---------------------------------------
    # Derived Features
    # ---------------------------------------
    daily["Avg Basket Size"] = (
        daily["Item Quantity"] /
        daily["Total Transactions"]
    )

    daily["Avg Basket Price"] = (
        daily["Total Sales"] /
        daily["Total Transactions"]
    )

    # ---------------------------------------
    # Add Channel
    # ---------------------------------------
    daily["Channel"] = "Combined"

    # ---------------------------------------
    # Reset Index
    # ---------------------------------------
    daily = daily.reset_index()

    # ---------------------------------------
    # Reorder Columns
    # ---------------------------------------
    daily = daily[
        [
            "transaction_date",
            "Total Sales",
            "Loyal Sales",
            "Non Loyal Sales",
            "Total Transactions",
            "Loyal Transactions",
            "Non Loyal Transactions",
            "Channel",
            "Item Quantity",
            "Avg Basket Size",
            "Avg Basket Price",
        ]
    ]

    return daily
```

```python
def add_time_features(df):
    """
    Add calendar-based features to the daily dataset.
    """

    data = df.copy()

    # ---------------------------------------
    # Ensure Date is datetime
    # ---------------------------------------
    data["transaction_date"] = pd.to_datetime(data["transaction_date"])

    # ---------------------------------------
    # Day of Week
    # Monday=1 ... Sunday=7
    # ---------------------------------------
    data["Day of Week"] = data["transaction_date"].dt.dayofweek + 1

    # ---------------------------------------
    # Week of Year
    # ---------------------------------------
    data["Week of Year"] = (
        data["transaction_date"]
        .dt.isocalendar()
        .week
        .astype(int)
    )

    # ---------------------------------------
    # Month
    # ---------------------------------------
    data["Month"] = data["transaction_date"].dt.month

    # ---------------------------------------
    # Weekend (Egypt)
    # Friday = 5
    # Saturday = 6
    # ---------------------------------------
    data["Is Weekend"] = (
        data["transaction_date"]
        .dt.dayofweek
        .isin([4, 5])
        .astype(int)
    )

    return data
```

```python
def get_parquet_files(data_path):
    """
    Return all parquet files sorted by name.
    """

    files = sorted(Path(data_path).glob("*.parquet"))

    print("=" * 60)
    print(f"Found {len(files)} parquet files")
    print("=" * 60)

    for file in files:
        print(file.name)

    return files
```

```python
def load_month(file_path):
    """
    Load one monthly parquet file.
    """

    print("=" * 60)
    print(f"Reading {Path(file_path).name}")
    print("=" * 60)

    df = pd.read_parquet(file_path)
    df = df.drop(columns=["Store No_", "Opening Date", "Source_File"], errors="ignore")

    print(f"Shape : {df.shape}")

    return df
```

```python
def preprocess_transactions(df):
    """
    Basic preprocessing.
    """

    data = df.copy()

    data["transaction_date"] = pd.to_datetime(data["transaction_date"])

    data["net_amount"] *= -1
    data["quantity"] *= -1

    data = data[
        (data["net_amount"] > 0) &
        (data["quantity"] > 0)
    ].copy()

    return data
```

```python
def aggregate_daily(df):
    """
    Aggregate transaction data into daily metrics.
    """

    daily = (

        df.groupby("transaction_date")

        .agg(

            Total_Sales=("net_amount","sum"),

            Loyal_Sales=(
                "net_amount",
                lambda x: x[df.loc[x.index,"is_loyal"]].sum()
            ),

            Non_Loyal_Sales=(
                "net_amount",
                lambda x: x[~df.loc[x.index,"is_loyal"]].sum()
            ),

            Total_Transactions=("receipt_no","nunique"),

            Loyal_Transactions=(
                "receipt_no",
                lambda x: x[df.loc[x.index,"is_loyal"]].nunique()
            ),

            Non_Loyal_Transactions=(
                "receipt_no",
                lambda x: x[~df.loc[x.index,"is_loyal"]].nunique()
            ),

            Item_Quantity=("quantity","sum")

        )

        .reset_index()

    )

    daily["Avg_Basket_Size"] = (
        daily["Item_Quantity"] /
        daily["Total_Transactions"]
    )

    daily["Avg_Basket_Price"] = (
        daily["Total_Sales"] /
        daily["Total_Transactions"]
    )

    return daily
```

```python
def process_month(file_path, output_path):
    """
    Process one monthly parquet file and save
    one daily parquet for each Store x Channel.
    """

    df = load_month(file_path)
    df = preprocess_transactions(df)

    stores = sorted(df["store_code"].dropna().unique())

    month_name = Path(file_path).stem.replace("transactions_", "")

    print(f"\nFound {len(stores)} Stores\n")

    for store in stores:

        print(f"Processing Store : {store}")

        store_df = df[df["store_code"] == store]

        store_folder = Path(output_path) / store
        store_folder.mkdir(parents=True, exist_ok=True)

        # -----------------------------
        # Individual Channels
        # -----------------------------

        for channel_name, folder_name in CHANNELS.items():

            channel_df = store_df[
                store_df["channel"] == channel_name
            ]

            if len(channel_df) == 0:
                continue

            daily = aggregate_daily(channel_df)
            daily = add_time_features(daily)

            save_folder = store_folder / folder_name
            save_folder.mkdir(exist_ok=True)

            save_path = save_folder / f"{month_name}.parquet"

            daily.to_parquet(save_path, index=False)

            print(f"   ✓ {folder_name}")

            del channel_df
            del daily
            gc.collect()

        # -----------------------------
        # Combined
        # -----------------------------

        combined = aggregate_daily(store_df)
        combined = add_time_features(combined)

        combined_folder = store_folder / "combined"
        combined_folder.mkdir(exist_ok=True)

        combined.to_parquet(
            combined_folder / f"{month_name}.parquet",
            index=False
        )

        print("   ✓ combined")

        del combined
        del store_df
        gc.collect()

    del df
    gc.collect()

    print(f"\nFinished : {month_name}")
```

```python
from pathlib import Path
import pandas as pd

SOURCE_PATH = "/content/drive/MyDrive/Spinneys/Forecasting/Processed_Data"
OUTPUT_PATH = "/content/drive/MyDrive/Spinneys/Forecasting/Train_Processed_Data"


def create_train_dataset(source_path, output_path):
    """
    Copy all processed files while removing January 2026.
    Folder structure is preserved.
    """

    source_path = Path(source_path)
    output_path = Path(output_path)

    files = [
        file
        for store_folder in Path(source_path).iterdir()
        if store_folder.is_dir()
        for file in store_folder.glob("*.parquet")
    ]
    print("=" * 60)
    print(f"Found {len(files)} parquet files")
    print("=" * 60)

    for file in files:

        # اقرأ الملف
        df = pd.read_parquet(file)

        df["transaction_date"] = pd.to_datetime(df["transaction_date"])

        before = len(df)

        # احذف يناير 2026
        df = df[
            ~(
                (df["transaction_date"].dt.year == 2026) &
                (df["transaction_date"].dt.month == 1)
            )
        ]

        after = len(df)

        # نفس الـ Path لكن داخل Train_Processed_Data
        relative_path = file.relative_to(source_path)

        save_path = output_path / relative_path

        save_path.parent.mkdir(parents=True, exist_ok=True)

        df.to_parquet(save_path, index=False)

        print(
            f"{relative_path} | Removed: {before-after}"
        )

    print("\nDone ✓")
```
